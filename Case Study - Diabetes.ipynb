```python
df_diabetes_clean = df_diabetes.copy()
df_diabetes_clean.head()
```

```python
df_diabetes_clean['Insulin_Missing'] = (df_diabetes_clean['Insulin']==0).astype(int)
```

```python
df_diabetes_clean.head()
```

```python
df_diabetes_clean['SkinThickness_Missing']= (df_diabetes_clean['SkinThickness']==0).astype(int)
df_diabetes_clean.head()
```

```python
cols = ['Glucose','BloodPressure','BMI','Insulin','SkinThickness']
```

```python
df_diabetes_clean[cols] = df_diabetes_clean[cols].replace(0, np.nan)
```

```python
df_diabetes_clean.head()
```

```python
for col in cols:
    df_diabetes_clean[col].fillna(df_diabetes_clean[col].median(), inplace=True)
```
