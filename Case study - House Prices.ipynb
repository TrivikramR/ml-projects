```python
df_house_prices=df_house_prices.copy()
```

```python
df_house_prices.drop(columns=['PoolQC','MiscFeature','Alley','Fence','MasVnrType'],inplace=True)
df_house_prices
```

```python
garage_cols=[['GarageType','GarageFinish','GarageQual','GarageCond']]
```

```python
for col in garage_cols:
    df_house_prices[col].fillna('None')
```

```python
df_house_prices.head(20)
```

```python
df_house_prices['FireplaceQu']=df_house_prices['FireplaceQu'].fillna('None')
```

```python
df_house_prices['FireplaceQu']
```

```python
df_house_prices['GarageYrBlt'] = df_house_prices['GarageYrBlt'].fillna(0)
```

```python
df_house_prices
```

```python
(df_house_prices['GarageYrBlt']==0).sum()
```

```python
bsmt_cols=[['BsmtFinType2','BsmtFinType1','BsmtExposure','BsmtCond','BsmtQual']]
```

```python
for col in bsmt_cols:
    df_house_prices[col].fillna('None')
```

```python
df_house_prices['MasVnrArea']= df_house_prices['MasVnrArea'].fillna(0)
```

```python
df_house_prices=df_house_prices.dropna(subset=['Electrical'])
```

```python
df_house_prices['LotFrontage']=df_house_prices.groupby('Neighborhood')['LotFrontage']\
.transform(lambda x:x.fillna(x.median()))
```
